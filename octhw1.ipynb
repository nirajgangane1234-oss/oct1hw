```python
# WAP to Reverse given string

s = "Niraj Rajesh Gangane"
reverse = s[::-1]
print("Reverse string",reverse)
```

```python
# WAP to check whether given string is Pallindrome or not...

string = "wow"
reverse = string [::-1]
if string == reverse:
    print("Given string is Pallindrome")
else:
    print("Given string is not Pallindrome")
```

```python
# WAP to count vowels and consonants from given string

string = "Niraj Rajesh Gangane"
vowels = 0
consonants = 0
for ch in string:
    if ch.lower() in "aeiou":
        vowels += 1
    elif ch.isalpha():
        consonants += 1
print("Vowels:",vowels) 
print("Consonants:",consonants)       
```

```python
# WAP to count number of in a string

string = "Niraj Rajesh Gangane"
words = string.split()
print("Number of words:", len(words))
```

```python
# WAP to count frequency of a character

string = "Niraj Rajesh Gangane"
char = "a"
count = string.count(char)
print("Frequency", count)
```

```python
# WAP to remove all spaces from a string

string = "Niraj Rajesh Gangane"
result = string.replace(" ","")
print("String without spaces:",result)
```
